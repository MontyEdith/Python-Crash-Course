# While loop

```python
# Pizza Topping

prompt = "toppings"
while prompt != "quit":
    toppings = input("Enter the toppings you want to add to your pizza: ")
    print(f"Adding {toppings} to your pizza")
    prompt = input(f"Once you are done, enter 'quit' to stop:")
```

```python
#Movie Tickets

a = True
while a:
    age = int(input("Enter your age: "))
    if age < 3:
        print("Ticket is free")
    elif age < 12:
        print("Ticket is $10")
    elif age > 12:
            print("Ticket is $15")
    a = False
```

```python
# Three Exits

while True:
    age = int(input("Enter your age: "))
    if age < 3:
        print("Ticket is free")
    elif age < 12:
        print("Ticket is $10")
    elif age > 12:
            print("Ticket is $15")

    choice = input("Do you want to check ticket for other person? (y/n)")

    if choice == "n":
        break
```

```python
#Deli

sandwitch_orders = ["cheese sandwich", "chicken sandwich", "green sandwich"]
finished_sandwich = []

while sandwitch_orders:
    current_sandwitch = sandwitch_orders.pop()

    print(f"I made you {current_sandwitch}")
    finished_sandwich.append(current_sandwitch)

print("\nFollowing sandwiches were made")
for sandwich in finished_sandwich:
    print(sandwich)
```

```python
# No Pastrami

sandwitch_orders = ["cheese sandwich", "pastrami sandwich", "chicken sandwich", "pastrami sandwich", "green sandwich", "pastrami sandwich"]
finished_sandwich = []
print("Deli has run out of pastrami")

while sandwitch_orders:
    current_sandwitch = sandwitch_orders.pop()

    if current_sandwitch != "pastrami sandwich":
        finished_sandwich.append(current_sandwitch)

print("\nFollowing sandwiches are left: ")
for sandwich in finished_sandwich:
    print(sandwich)
```
