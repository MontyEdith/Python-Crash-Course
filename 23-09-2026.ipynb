## Modifying,Adding,andRemovingElements

```python
#Modifying elements in a list

fav_mode_transportation = ["Lamborgini", "Bronco", "Mercedes Benz S Class", "Triumph"]
print(fav_mode_transportation)

fav_mode_transportation[2] = "G-Wagon"
print(fav_mode_transportation)
```

```python
#Adding elements to a list
#Append

fav_mode_transportation = ["Lamborgini", "Bronco", "Mercedes Benz S Class", "Triumph"]
print(fav_mode_transportation)

fav_mode_transportation.append("Audi")
print(fav_mode_transportation)
```

```python
fav_mode_transportation = []
print(fav_mode_transportation)

fav_mode_transportation.append("Lamborgini")
fav_mode_transportation.append("Bronco")
fav_mode_transportation.append("Mercedes Benz S Class")
fav_mode_transportation.append("Triumph")
print(fav_mode_transportation)
```

```python
#Insert
fav_mode_transportation = ["Lamborgini", "Bronco", "Mercedes Benz S Class", "Triumph"]
print(fav_mode_transportation)

fav_mode_transportation.insert(2,"Audi")
print(fav_mode_transportation)
```

```python
#Removing elements from a list
#del statement

fav_mode_transportation = ["Lamborgini", "Bronco", "Mercedes Benz S Class", "Triumph"]
print(fav_mode_transportation)

del fav_mode_transportation[2]
print(fav_mode_transportation)
```

```python
#pop() method

fav_mode_transportation = ["Lamborgini", "Bronco", "Mercedes Benz S Class", "Triumph"]
print(fav_mode_transportation)

popped_transportation = fav_mode_transportation.pop()
print(fav_mode_transportation)
print(popped_transportation)
```

```python
#Organizing a List
#sort() method

guest = ['Bhuvan bam', 'Mahatama Gandhi', 'Monty Python', 'Bhagat Singh', 'APJ Abdul kalam', 'RDJ', 'SRK', 'Albert einstein']
print(guest)

guest.sort()
print(guest)
```

```python
#sorted() Function

guest = ['Bhuvan bam', 'Mahatama Gandhi', 'Monty Python', 'Bhagat Singh', 'APJ Abdul kalam', 'RDJ', 'SRK', 'Albert einstein']

print("Here is the original list: ")
print(guest)
print("\n")
print("Here is the sorted list: ")
print(sorted(guest))
print("\n")
print("Here is the original list again: ")
print(guest)
```

```python
#reverse() method

guest = ['Bhuvan bam', 'Mahatama Gandhi', 'Monty Python', 'Bhagat Singh', 'APJ Abdul kalam', 'RDJ', 'SRK', 'Albert einstein']
print(guest)

guest.reverse()
print(guest)
```

```python
#len() function

print(guest)

print(len(guest))
```

```python
#Seeing the World

places_to_visit = ["Germany","Amsterdam","Paris","Bhutan","Thailand"]

print(f"Original order of the list: {places_to_visit}")
print("\n")

print(sorted(places_to_visit))
print(f"Order is still the same: {places_to_visit}")
print("\n")

print(sorted(places_to_visit, reverse=True))
print(f"Order is still the same: {places_to_visit}")
print("\n")

places_to_visit.reverse()
print(f"Order has been changed and now it's in reverse order: {places_to_visit}")
print("\n")

places_to_visit.reverse()
print(f"Order has been again changed and now it's in original order: {places_to_visit}")
print("\n")

places_to_visit.sort()
print(f"Order has been changed and now it's in alphabetical order: {places_to_visit}")
print("\n")

places_to_visit.sort(reverse=True)
print(f"Order has been changed and now it's in reverse-alphabetical order: {places_to_visit}")
print("\n")
```
