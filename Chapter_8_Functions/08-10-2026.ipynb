```python
#Album

def make_album(artist_name, album_title, songs=None):
    if songs:
        return f"Artist: {artist_name.title()}, Album: {album_title.title()}, Songs: {songs}\n"
    else:
        return f"Artist: {artist_name.title()}, Album: {album_title.title()}\n"

music_album1 = make_album("sonu nigam", "tu", 16)
print(music_album1)

music_album2 = make_album("shaan", "chand")
print(music_album2)

music_album3 = make_album("arijit singh", "pal", 7)
print(music_album3)
```

```python
def make_pizza(Unprepared_pizza, completed_orders):

    while Unprepared_pizza:
        current_order = Unprepared_pizza.pop()
        print(f"Preparing {current_order}")
        completed_orders.append(current_order)

    print("\nFollowing are the completed orders: ")
    for i in completed_orders:
        print(i)

    return completed_orders

pizza = make_pizza(["Farmhouse Pizza", "Green Valley Pizza", "Margerita Pizza"], ["Maccaroni Pizza"])
print(pizza)
```

```python
#Sending Messages

def sent_messages(message_list,new_message_list):
    while message_list:
        current_message = message_list.pop()
        print(f"Sending {current_message}")
        new_message_list.append(current_message)

    return f"\nOld message list : {message_list} \nNew message list : {new_message_list}"

sending_messages = sent_messages(["Hello","Good morning", "Toodles"],[])
print(sending_messages)
```

```python
#Archived Messages

def show_messages(messages):
    """Print all messages in the list."""
    print("Showing all messages:")
    for message in messages:
        print(message)

def send_messages(messages, sent_messages):
    """Print each message, and then move it to sent_messages."""
    print("\nSending all messages:")
    while messages:
        current_message = messages.pop()
        print(current_message)
        sent_messages.append(current_message)

messages = ["hello there", "how are u?", ":)"]
show_messages(messages)

sent_messages = []
send_messages(messages[:], sent_messages)

print("\nFinal lists:")
print(messages)
print(sent_messages)
```
