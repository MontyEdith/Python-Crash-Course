## Nesting

```python
users = {
    "ronny" : {
        "first_name" : "rishabh",
        "last_name" : "singh",
        "location" : "Bangalore" 
    },

    "edith" : {
        "first_name" : "barry",
        "last_name" : "allen",
        "location" : "star city" 
    }
}

for usename,user_info in users.items():
    print(f"{usename.title()}")
    full_name = f"{user_info["first_name"]} {user_info["last_name"]}"
    location = f"{user_info["location"]}"

    print(f"Full name: {full_name.title()}")
    print(f"Lacation: {location.title()}\n")
```

```python
#People

list = [
    {
    "first_name" : "rishabh",
    "last_name" : "singh",
    "age" : 31,
    "city" : "bangalore"
    },

    {
    "first_name" : "barry",
    "last_name" : "allen",
    "age" : 24,
    "city" : "star city"
    },

    {
    "first_name" : "bruce",
    "last_name" : "wayne",
    "age" : 45,
    "city" : "gotham"
    }
]

for i in list:
    for key,value in i.items():
        full_name = f"{i["first_name"]} {i["last_name"]}"
        age = i["age"]
        location = i["city"]

    print(f"Full Name: {full_name.title()}")
    print(f"Age: {age}")
    print(f"Location: {location.title()}\n")
```
